```python
from innolux import *
data_root = data_path / Path("T2_6638_ZIX-7061XX_TR35")
```

```python
df = pl.read_excel(data_root / "20260707_第一批bonding_均勻性量測.xlsx")
df
```

```python
from matplotlib import transforms

fig, ax = plt.subplots(figsize=(8,6))
sns.boxplot(
    data=df,
    x="條件",
    y="Vcom 均勻性",
    ax=ax,
    hue="條件",
    width=0.4,
)

sns.stripplot(
    data=df,
    x="條件",
    y="Vcom 均勻性",
    ax=ax,
    jitter=0.1,
    hue="條件",
)

shift_right = transforms.ScaledTranslation(0.7, 0, fig.dpi_scale_trans)
for collections in ax.collections:
    collections.set_transform(collections.get_transform() + shift_right)

ax.axhline(y=5.0, color="red", linestyle="--", linewidth=2)
ax.set_xbound(upper=3)
```
